```python
import pandas as pd
import numpy as np

from scipy.stats import spearmanr, kendalltau
```

```python
# =========================
# CONFIGURATION
# =========================

INPUT_FILE = "phase2_results.xlsx"
OUTPUT_FILE = "phase3_results.xlsx"

MODEL_COLUMN = "Model"
CONSENSUS_COLUMN = "Consensus_Rank"

RANK_COLUMNS = [
    "MARCOS",
    "SECA",
    "MUTRISS",
    "MAIRCA",
    "CODAS",
    "RAFSI",
    "CRADIS",
    "CoCoSo",
    "MABAC",
    "COPRAS",
    "RAM",
    "WASPAS",
    "PIV",
    "WISP"
]
```

```python
df = pd.read_excel(INPUT_FILE)

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
# ============================
# Validate Required Columns
# ============================

MODEL_COLUMN = "Model"

CONSENSUS_COLUMN = "Overall_Consensus_Rank"

RANK_COLUMNS = [
    "MARCOS",
    "SECA",
    "MUTRISS",
    "MAIRCA",
    "CODAS",
    "RAFSI",
    "CRADIS",
    "CoCoSo",
    "MABAC",
    "COPRAS",
    "RAM",
    "WASPAS",
    "PIV",
    "WISP"
]

required_columns = [MODEL_COLUMN, CONSENSUS_COLUMN] + RANK_COLUMNS

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

print("✅ Data validation successful.")
```

```python
#step 10

spearman_matrix = pd.DataFrame(
    index=RANK_COLUMNS,
    columns=RANK_COLUMNS,
    dtype=float
)

for method1 in RANK_COLUMNS:
    for method2 in RANK_COLUMNS:

        correlation, _ = spearmanr(
            df[method1],
            df[method2]
        )

        spearman_matrix.loc[
            method1,
            method2
        ] = correlation

print("Spearman Correlation Matrix:")
display(spearman_matrix)
```

```python
# step 11
kendall_matrix = pd.DataFrame(
    index=RANK_COLUMNS,
    columns=RANK_COLUMNS,
    dtype=float
)

for method1 in RANK_COLUMNS:
    for method2 in RANK_COLUMNS:

        tau, _ = kendalltau(
            df[method1],
            df[method2]
        )

        kendall_matrix.loc[
            method1,
            method2
        ] = tau

print("Kendall's Tau Matrix:")
display(kendall_matrix)
```

```python
#step 12
def calculate_top_k_agreement(
    data,
    method_column,
    consensus_column,
    model_column,
    k
):
    consensus_top_k = set(
        data.sort_values(consensus_column)
            .head(k)[model_column]
    )

    method_top_k = set(
        data.sort_values(method_column)
            .head(k)[model_column]
    )

    overlap = len(
        consensus_top_k.intersection(method_top_k)
    )

    agreement = overlap / k

    return overlap, agreement
```

```python
top_k_results = []

for method in RANK_COLUMNS:

    overlap_10, agreement_10 = calculate_top_k_agreement(
        df,
        method,
        CONSENSUS_COLUMN,
        MODEL_COLUMN,
        10
    )

    overlap_20, agreement_20 = calculate_top_k_agreement(
        df,
        method,
        CONSENSUS_COLUMN,
        MODEL_COLUMN,
        20
    )

    top_k_results.append({
        "Method": method,
        "Top10_Overlap": overlap_10,
        "Top10_Agreement_%": agreement_10 * 100,
        "Top20_Overlap": overlap_20,
        "Top20_Agreement_%": agreement_20 * 100
    })

top_k_df = pd.DataFrame(top_k_results)

display(top_k_df)
```

```python
# step 13 - rank stability
rank_stability_df = df[
    [MODEL_COLUMN] + RANK_COLUMNS
].copy()

rank_stability_df["Rank_Stability_SD"] = (
    rank_stability_df[RANK_COLUMNS].std(axis=1)
)

rank_stability_df = rank_stability_df.sort_values(
    "Rank_Stability_SD"
)

display(
    rank_stability_df[
        [MODEL_COLUMN, "Rank_Stability_SD"]
    ]
)
```

```python
print("Minimum SD :", rank_stability_df["Rank_Stability_SD"].min())
print("Maximum SD :", rank_stability_df["Rank_Stability_SD"].max())
print("Average SD :", rank_stability_df["Rank_Stability_SD"].mean())
```

```python
def stability_category(sd):
    if sd <= 25:
        return "High Stability"
    elif sd <= 75:
        return "Moderate Stability"
    else:
        return "Low Stability"

rank_stability_df["Stability_Category"] = (
    rank_stability_df["Rank_Stability_SD"]
    .apply(stability_category)
)

display(rank_stability_df[
    ["Model", "Rank_Stability_SD", "Stability_Category"]
].head(20))
```

```python
#step 14 - Method Representative Rank

representative_results = []

for method in RANK_COLUMNS:

    other_methods = [
        m for m in RANK_COLUMNS
        if m != method
    ]

    average_correlation = (
        spearman_matrix
        .loc[method, other_methods]
        .mean()
    )

    representative_results.append({
        "Method": method,
        "Average_Spearman_Correlation":
            average_correlation
    })

representative_df = pd.DataFrame(
    representative_results
)
```

```python
representative_df[
    "Representative_Rank"
] = representative_df[
    "Average_Spearman_Correlation"
].rank(
    ascending=False,
    method="min"
)

representative_df = representative_df.sort_values(
    "Representative_Rank"
)

display(representative_df)
```

```python
#final summary

average_spearman = (
    spearman_matrix
    .replace(1.0, np.nan)
    .mean(axis=1)
)

average_kendall = (
    kendall_matrix
    .replace(1.0, np.nan)
    .mean(axis=1)
)
```

```python
phase3_summary = representative_df.copy()

phase3_summary["Average_Spearman"] = (
    phase3_summary["Method"]
    .map(average_spearman)
)

phase3_summary["Average_Kendall"] = (
    phase3_summary["Method"]
    .map(average_kendall)
)

phase3_summary = phase3_summary.merge(
    top_k_df[
        [
            "Method",
            "Top10_Agreement_%",
            "Top20_Agreement_%"
        ]
    ],
    on="Method"
)

display(phase3_summary)
```

```python
with pd.ExcelWriter(OUTPUT_FILE) as writer:

    spearman_matrix.to_excel(
        writer,
        sheet_name="Spearman"
    )

    kendall_matrix.to_excel(
        writer,
        sheet_name="Kendall"
    )

    top_k_df.to_excel(
        writer,
        sheet_name="Top_K_Agreement",
        index=False
    )

    rank_stability_df.to_excel(
        writer,
        sheet_name="Rank_Stability",
        index=False
    )

    representative_df.to_excel(
        writer,
        sheet_name="Representative_Method",
        index=False
    )

    phase3_summary.to_excel(
        writer,
        sheet_name="Phase3_Summary",
        index=False
    )

print(f"Phase 3 results saved to: {OUTPUT_FILE}")
```
